```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv("NFLX.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df = df.drop(["Open","High","Low","Close","Volume"],axis=1)
df
```

```python
df = df.groupby("Date")['Adj Close'].sum().reset_index()
# Group the DataFrame by. 'Date' column, calculate the sum of 'Adj .Close' values for each date,
# and reset the index to obtain a new DataFrame with 'Date' and 'Adj Close' columns.
```

```python
df
```

```python
df.Date = pd.to_datetime(df.Date) # Convert the 'Date' column to datetime format df['Date' ] == df.Date
df.head()
```

```python
df.set_index(["Date"],inplace=True)
df.head()
```
