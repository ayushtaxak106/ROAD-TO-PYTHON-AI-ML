# MULTI LAYER PERCENPTRON ARCHITECTURE 
INPUT LAYER --> HIDDEN LAYER (1) --> OUTPUT LAYER

BUILD MODEL

```python
class Neuralnetwork:
    def __init__(self,input_size,hidden_size,output_size,learning_rate = 0.01,epochs = 100):
        self.input_size = input_size #no of neurons as per the number of features  in the layer
        self.hidden_size = hidden_size #no. of neurons in the hidden layer layer (1st HL)
        self.output_size = output_size #no of neurons on the output layer
        self.learning_rate = learning_rate #to set the user  defined learning rate  for the gradient descent; default is set to 0.01
        self.epochs = epochs
```
