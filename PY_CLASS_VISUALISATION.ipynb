```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data1 = pd.read_csv('cars.csv',skipinitialspace=True)
```

```python
data1
```

#MPG--> miles per gallon
This shows the fuel efficiency of the car

#cylinders -->
Number of cylinders in the car engine(eg:4,6,8)

#cubicinches-->
how much air fuel mixture the engine can burn.

#hp(horse power)-->
power output of the engine

#weightlbs-->
car's weight in pounds

#time-to-60-->
the time  in seconds car takes to accelerate from 0 to 60 mph

#year-->
 in which year car was manufactured.

#brand-->
brand of the car.

```python
fig,ax = plt.subplots(2,2,figsize=(20,10))

plt.subplot(2,2,1)
sns.scatterplot(x=data1['mpg'],y=data1['hp'],data=data1,hue='brand')
plt.xlabel('miles per gallon')
plt.ylabel('horse power')
plt.title('mpg vs hp')

plt.subplot(2,2,2)
sns.scatterplot(x=data1['mpg'],y=data1['weightlbs'],data=data1,hue='brand')
plt.xlabel('miles per gallon')
plt.ylabel('weightlbs')
plt.title('mpg vs weightlbs')

plt.subplot(2,2,3)
sns.scatterplot(x=data1['hp'],y=data1['mpg'],data=data1,hue='brand')
plt.xlabel('hp')
plt.ylabel('mpg')
plt.title('hp vs mpg')

plt.subplot(2,2,4)
sns.scatterplot(x=data1['hp'],y=data1['time-to-60'],data=data1,hue='brand')
plt.xlabel('hp')
plt.ylabel('time-to-60')
plt.title('hp vs time-to-60')

plt.show()

```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.barplot(data=data1, x="brand", y="mpg", hue="cylinders",ax=axes[0, 0])
axes[0, 0].set_title("MPG by Brand")

sns.barplot(data=data1, x="brand", y="hp", hue="cylinders",ax=axes[0, 1])
axes[0, 1].set_title("HP by Brand")

sns.barplot(data=data1, x="brand", y="weightlbs", hue="cylinders",ax=axes[1, 0])
axes[1, 0].set_title("Weight by Brand")

sns.barplot(data=data1, x="brand", y="time-to-60", hue="cylinders",ax=axes[1, 1])
axes[1, 1].set_title("time-to-60 by Brand")
plt.show()
```

```python
fig = plt.subplots(2,2,figsize=(20,10))

plt.subplot(2,2,1)
sns.boxplot(x=data1['cylinders'],y=data1['hp'],data=data1,hue=data1['brand'])
plt.xlabel('cylinders')
plt.ylabel('hp')
plt.title('cylinders vs hp')

plt.subplot(2,2,2)
sns.boxplot(x=data1['cylinders'],y=data1['mpg'],data=data1,hue=data1['brand'])
plt.xlabel('cylinders')
plt.ylabel('mpg')
plt.title('cylinders vs mpg')

plt.subplot(2,2,3)
sns.boxplot(x=data1['cylinders'],y=data1['weightlbs'],data=data1,hue=data1['brand'])
plt.xlabel('cylinders')
plt.ylabel('weightlbs')
plt.title('cylinders vs weightlbs')

plt.subplot(2,2,4)
sns.boxplot(x=data1['cylinders'],y=data1['time-to-60'],data=data1,hue=data1['brand'])
plt.xlabel('cylinders')
plt.ylabel('time-to-60')
plt.title('cylinders vs time-to-60')

plt.show()
```
